## Problem 1: Theoretical Derivations in Unsupervised Deep Learning

In this problem we derive several key theoretical results:

### (1a) Optimal Discriminator in GANs

The GAN objective is given by

$$V(G,D)= \int_x p_{data}(x)\log(D(x))dx + \int_z p_z(z)\log(1-D(G(z)))dz.$$ 

Assuming that at optimum we have $p_g = p_{data}$, we can write the value function at each $x$ as

$$V(D(x)) = p_{data}(x) \log(D(x)) + p_{data}(x) \log(1-D(x)) = p_{data}(x)\left[\log(D(x)(1-D(x))\right].$$

This is maximized when $D(x)(1-D(x))$ is maximized, which occurs at $D(x)=\frac{1}{2}$. 

Substituting $D(x)=\frac{1}{2}$ into the value function gives

$$V(G,D^*) = \int_x p_{data}(x) \log\left(\frac{1}{2}\right) dx + \int_z p_z(z) \log\left(\frac{1}{2}\right) dz = \log\left(\frac{1}{2}\right),$$

and hence the optimum value is

$$V(G,D^*) = -\log(4),$$

which completes the derivation.

### (1b) The Reparameterization Trick in VAEs

In Variational Autoencoders (VAEs), we sample the latent variable $z$ using the reparameterization trick:

$$z = \mu_{z|x} + \sigma_{z|x}\epsilon, \quad \epsilon \sim \mathcal{N}(0,1).$$

Since $\epsilon$ is drawn from a standard normal distribution, it follows that $z$ is normally distributed with parameters $\mu_{z|x}$ and $\sigma_{z|x}$, i.e.,

$$z \sim \mathcal{N}(\mu_{z|x}, \sigma_{z|x}^2).$$

This trick enables backpropagation through the stochastic sampling step.

### (1c) KL Divergence between Two Bernoulli Distributions

Let $p(x) = \text{Bern}(p_1)$ and $q(x) = \text{Bern}(p_2)$. The KL divergence is defined as

$$\text{KL}(p\|q) = \sum_{x \in \{0,1\}} p(x) \log \frac{p(x)}{q(x)}.$$ 

Evaluating the sum gives

$$\text{KL}(p\|q) = p_1 \log \frac{p_1}{p_2} + (1-p_1) \log \frac{1-p_1}{1-p_2}.$$ 

This result is used in VAEs to regularize the latent space by measuring the divergence between the approximate posterior and the prior, thereby ensuring that the latent representations do not deviate too much from a normal distribution.

## Problem 2: Image Classification for Planes, Ships, and Trucks

In this problem we develop an image classifier to automatically classify images of planes, ships, and trucks. 

### (2a) Model Architecture Choice

We choose a VGG-11 architecture with Batch Normalization (BN) after the convolutional layers. This architecture is a good start because:

- **Simplicity:** VGG networks are simple and effective.
- **BN:** Batch Normalization improves training stability and convergence.
- **Customization:** We will implement our own version (without pretrained weights) to ensure a full understanding of the process.

### (2b) Explanation of Dropout

Dropout is a regularization technique where, during training, a fraction of the neurons is randomly set to zero. This prevents the network from overfitting and encourages redundancy in representations. In our network, we will use a dropout rate of 0.5 after each fully connected layer with 4096 filters.

### (2c) Implementing the VGG-11 with BN and Dropout

Below we implement our custom VGG-11 model using PyTorch.

```python
import torch
import torch.nn as nn
import torch.nn.functional as F
import sys
import os

# Add the src folder to the Python path
sys.path.append(os.path.join(os.getcwd(), 'src/'))

class VGG11Custom(nn.Module):
    def __init__(self, num_classes=3):
        super(VGG11Custom, self).__init__()
        self.features = nn.Sequential(
            # Conv Block 1
            nn.Conv2d(3, 64, kernel_size=3, padding=1),
            nn.BatchNorm2d(64),
            nn.ReLU(inplace=True),
            nn.MaxPool2d(kernel_size=2, stride=2),

            # Conv Block 2
            nn.Conv2d(64, 128, kernel_size=3, padding=1),
            nn.BatchNorm2d(128),
            nn.ReLU(inplace=True),
            nn.MaxPool2d(kernel_size=2, stride=2),

            # Conv Block 3
            nn.Conv2d(128, 256, kernel_size=3, padding=1),
            nn.BatchNorm2d(256),
            nn.ReLU(inplace=True),
            nn.Conv2d(256, 256, kernel_size=3, padding=1),
            nn.BatchNorm2d(256),
            nn.ReLU(inplace=True),
            nn.MaxPool2d(kernel_size=2, stride=2),

            # Conv Block 4
            nn.Conv2d(256, 512, kernel_size=3, padding=1),
            nn.BatchNorm2d(512),
            nn.ReLU(inplace=True),
            nn.Conv2d(512, 512, kernel_size=3, padding=1),
            nn.BatchNorm2d(512),
            nn.ReLU(inplace=True),
            nn.MaxPool2d(kernel_size=2, stride=2),

            # Conv Block 5
            nn.Conv2d(512, 512, kernel_size=3, padding=1),
            nn.BatchNorm2d(512),
            nn.ReLU(inplace=True),
            nn.Conv2d(512, 512, kernel_size=3, padding=1),
            nn.BatchNorm2d(512),
            nn.ReLU(inplace=True),
            nn.MaxPool2d(kernel_size=2, stride=2)
        )

        self.classifier = nn.Sequential(
            nn.Linear(512 * 3 * 3, 4096),
            nn.ReLU(inplace=True),
            nn.Dropout(0.5),  # Dropout with rate 0.5
            nn.Linear(4096, 4096),
            nn.ReLU(inplace=True),
            nn.Dropout(0.5),  # Dropout with rate 0.5
            nn.Linear(4096, num_classes)
        )

    def forward(self, x):
        x = self.features(x)
        x = x.view(x.size(0), -1)
        x = self.classifier(x)
        return x

# Instantiate the model
model = VGG11Custom(num_classes=3)
print(model)
```

### (2c) Training the VGG-11 Model

Below is a sample training loop. The training and evaluation data are loaded from the provided NPZ files. 



```python
import numpy as np
import matplotlib.pyplot as plt
from torch.utils.data import Dataset, DataLoader

class ImageDataset(Dataset):
    def __init__(self, images, labels=None, transform=None):
        self.images = images
        self.labels = labels
        self.transform = transform

    def __len__(self):
        return self.images.shape[0]

    def __getitem__(self, idx):
        image = self.images[idx]
        if self.transform:
            image = self.transform(image)
        if self.labels is not None:
            label = self.labels[idx]
            return image, label
        else:
            return image

# Load training and evaluation data
training_data = np.load ('data/problem2/training_data.npz')
train_images = training_data['a']  # shape: (1500, 96, 96, 3)
train_labels = training_data['b']

evaluation_data = np.load(training_data = np.load ("data/problem2/evaluation_data.npz")
eval_images = evaluation_data['a']  # shape: (2400, 96, 96, 3)
eval_labels = evaluation_data['b']

# Preprocess: Convert images to float32 and transpose to (C, H, W)
def preprocess(image):
    image = image.astype(np.float32) / 255.0
    # transpose (H, W, C) -> (C, H, W)
    image = np.transpose(image, (0, 1, 2))
    return torch.tensor(image)

from torchvision import transforms
transform = transforms.Lambda(lambda img: preprocess(img))

train_dataset = ImageDataset(train_images, train_labels, transform=transform)
eval_dataset = ImageDataset(eval_images, eval_labels, transform=transform)

# Create DataLoaders
train_loader = DataLoader(train_dataset, batch_size=32, shuffle=True)
eval_loader = DataLoader(eval_dataset, batch_size=32, shuffle=False)

# Set up training parameters
device = torch.device('cuda' if torch.cuda.is_available() else 'cpu')
model.to(device)
optimizer = torch.optim.Adam(model.parameters(), lr=1e-4)
criterion = nn.CrossEntropyLoss()

num_epochs = 10

# Initialize lists to store training loss and validation accuracy
train_losses = []
val_accuracies = []

for epoch in range(num_epochs):
    model.train()
    running_loss = 0.0
    for images, labels in train_loader:
        images = images.to(device)
        labels = labels.to(device).long()
        optimizer.zero_grad()
        outputs = model(images)
        loss = criterion(outputs, labels)
        loss.backward()
        optimizer.step()
        running_loss += loss.item() * images.size(0)
    epoch_loss = running_loss / len(train_loader.dataset)
    train_losses.append(epoch_loss)  # Store training loss
    print(f"Epoch {epoch+1}/{num_epochs}, Loss: {epoch_loss:.4f}")

    # Evaluate on validation set
    model.eval()
    correct = 0
    total = 0
    with torch.no_grad():
        for images, labels in eval_loader:
            images = images.to(device)
            labels = labels.to(device).long()
            outputs = model(images)
            _, predicted = torch.max(outputs, 1)
            total += labels.size(0)
            correct += (predicted == labels).sum().item()
    val_acc = correct / total * 100
    val_accuracies.append(val_acc)  # Store validation
    print(f"Validation Accuracy: {val_acc:.2f}%")

# Plot training loss and validation accuracy
plt.figure(figsize=(12, 5))

# Plot training loss
plt.subplot(1, 2, 1)
plt.plot(range(1, num_epochs + 1), train_losses, marker='o', label='Training Loss')
plt.xlabel('Epoch')
plt.ylabel('Loss')
plt.title('Training Loss Over Epochs')
plt.grid(True)
plt.legend()

# Plot validation accuracy
plt.subplot(1, 2, 2)
plt.plot(range(1, num_epochs + 1), val_accuracies, marker='o', label='Validation Accuracy')
plt.xlabel('Epoch')
plt.ylabel('Accuracy (%)')
plt.title('Validation Accuracy Over Epochs')
plt.grid(True)
plt.legend()

plt.tight_layout()
plt.show()
```

### (2d) Uncertainty Estimation using Dropout

To filter out images of an unwanted class (e.g., birds), we leverage dropout during inference to obtain uncertainty estimates. 

For each labeled image in the new dataset, we perform 10 forward passes with dropout enabled, compute the mean softmax output $\hat{y}$, and then calculate the entropy:

$$H(\hat{y}) = -\sum_{i=1}^{N_c} \hat{y}_i \log \hat{y}_i.$$ 

A higher entropy indicates greater uncertainty. We then plot histograms of the entropy for known and unwanted classes.

```python
def get_entropy(probs):
    # probs: tensor of shape (num_classes,)
    return -torch.sum(probs * torch.log(probs + 1e-8)).item()

def mc_dropout_inference(model, image, T=10):
    model.train()  # Enable dropout at inference time
    softmaxes = []
    with torch.no_grad():
        for _ in range(T):
            output = model(image.unsqueeze(0).to(device))
            softmax = F.softmax(output, dim=1).cpu().squeeze()
            softmaxes.append(softmax)
    mean_softmax = torch.stack(softmaxes, dim=0).mean(dim=0)
    return mean_softmax, get_entropy(mean_softmax)

# Load new evaluation data with labels
new_eval_data = np.load("data/problem2/new_evaluation_data_with_labels.npz")
new_eval_ita['a']
new_eval_labels = new_eval_data['b']

# Preprocess new evaluation images
new_eval_images_tensor = torch.stack([preprocess(img) for img in new_eval_images])

# Compute entropy for each image
entropies = []
for img in new_eval_images_tensor:
    _, entropy = mc_dropout_inference(model, img)
    entropies.append(entropy)

import matplotlib.pyplot as plt
plt.figure(figsize=(8, 6))
plt.hist(entropies, bins=30, color='blue', alpha=0.7)
plt.xlabel('Entropy')
plt.ylabel('Frequency')
plt.title('Histogram of Entropy for New Evaluation Data')
plt.show()

# In a real experiment, separate histograms for known and unwanted classes would be computed
# by comparing entropies based on the provided labels.
```

### (2e) Threshold Selection for Filtering Unwanted Images

Using the entropy estimates from (2d), we select a threshold to filter out images with high uncertainty (likely belonging to the unwanted class). 

Below, we count the number of unwanted images detected and the number of desired images that would be removed.

```python
# For illustration, assume we set a threshold on entropy
entropy_threshold = 0.025  # Example threshold

new_eval_labels = new_eval_labels.astype(int)

# Separate entropies by known and unwanted classes (assume label 3 corresponds to the unwanted class 'bird')
entropies_known = [e for e, l in zip(entropies, new_eval_labels) if l != 3]
entropies_unwanted = [e for e, l in zip(entropies, new_eval_labels) if l == 3]


# Plot histograms of entropies for known and unwanted classes
plt.figure(figsize=(10, 6))
plt.hist(entropies_known, bins=30, alpha=0.7, label='Known Classes', color='blue')
plt.hist(entropies_unwanted, bins=30, alpha=0.7, label='Unwanted Class', color='red')
plt.axvline(entropy_threshold, color='black', linestyle='--', label='Threshold')
plt.xlabel('Entropy')
plt.ylabel('Frequency')
plt.title('Entropy Distribution for Known and Unwanted Classes')
plt.legend()
plt.grid(True)
plt.show()
```

### (3b) Training a ResNet-18 Classifier

We now train a ResNet-18 model (without pretrained weights) on the traffic sign dataset. The training data (from the "train" folder) is split 80/20 into training and validation sets while preserving class distribution.

```python
import torchvision
import torchvision.transforms as transforms
from torch.utils.data import random_split

# Assume the traffic sign dataset is organized in folders
data_transform = transforms.Compose([
    transforms.Resize((96, 96)),
    transforms.ToTensor()
])

train_dataset = torchvision.datasets.ImageFolder(root="C:/Users/tihas/OneDrive - UiT Office 365/A_UIT/FYS-3033/HOME_EXAM/files/problem2/train", transform=data_transform)

# Split dataset (80% train, 20% validation)
train_size = int(0.8 * len(train_dataset))
val_size = len(train_dataset) - train_size
train_subset, val_subset = random_split(train_dataset, [train_size, val_size])

train_loader = DataLoader(train_subset, batch_size=32, shuffle=True)
val_loader = DataLoader(val_subset, batch_size=32, shuffle=False)

# Use torchvision's ResNet-18 (without pretrained weights)
from torchvision.models import resnet18
import torch.nn as nn

model_resnet = resnet18(pretrained=False)
num_ftrs = model_resnet.fc.in_features
model_resnet.fc = nn.Linear(num_ftrs, len(train_dataset.classes))

model_resnet.to(device)

optimizer = torch.optim.Adam(model_resnet.parameters(), lr=1e-4)
criterion = nn.CrossEntropyLoss()

num_epochs = 15
for epoch in range(num_epochs):
    model_resnet.train()
    running_loss = 0.0
    for images, labels in train_loader:
        images, labels = images.to(device), labels.to(device)
        optimizer.zero_grad()
        outputs = model_resnet(images)
        loss = criterion(outputs, labels)
        loss.backward()
        optimizer.step()
        running_loss += loss.item() * images.size(0)
    epoch_loss = running_loss / len(train_loader.dataset)
    print(f"Epoch {epoch+1}/{num_epochs}, Loss: {epoch_loss:.4f}")

    # Validation accuracy
    model_resnet.eval()
    correct = 0
    total = 0
    with torch.no_grad():
        for images, labels in val_loader:
            images, labels = images.to(device), labels.to(device)
            outputs = model_resnet(images)
            _, preds = torch.max(outputs, 1)
            total += labels.size(0)
            correct += (preds == labels).sum().item()
    print(f"Validation Accuracy: {100 * correct / total:.2f}%")
```

### (3d) Class Saliency Maps for the Worst-performing Class

For the class where our model fails on the test set, we generate class saliency maps (as described in [3]) for 10 training images to understand which image regions contribute most to the prediction.

```python
def compute_saliency(model, image, target_class):
    model.eval()
    image = image.unsqueeze(0).to(device)
    image.requires_grad_()
    output = model(image)
    score = output[0, target_class]
    score.backward()
    saliency, _ = torch.max(image.grad.data.abs(), dim=1)
    return saliency.squeeze().cpu().numpy()

# Select 10 images from the training set for the worst-performing class (e.g., class index 2)
worst_class = 2
import random
indices = random.sample(range(len(train_dataset)), 10)

import matplotlib.pyplot as plt
fig, axes = plt.subplots(2, 5, figsize=(15, 6))
for ax, idx in zip(axes.flatten(), indices):
    img, label = train_dataset[idx]
    saliency = compute_saliency(model_resnet, img, worst_class)
    ax.imshow(saliency, cmap='hot')
    ax.axis('off')
plt.suptitle('Class Saliency Maps for Worst-performing Class')
plt.tight_layout()
plt.show()
```

### (3e) Occlusion Analysis

For the same 10 training images, we perform occlusion analysis by occluding a 10x10 region in the image and monitoring changes in the classifier output. This analysis helps us understand which parts of the image are critical for the classification decision.

```python
def occlusion_analysis(model, image, target_class, occlusion_size=10):
    model.eval()
    image = image.unsqueeze(0).to(device)
    base_output = F.softmax(model(image), dim=1)[0, target_class].item()
    h, w = image.shape[2], image.shape[3]
    occlusion_map = np.zeros((h, w))
    
    for i in range(0, h, occlusion_size):
        for j in range(0, w, occlusion_size):
            occluded_image = image.clone()
            occluded_image[:, :, i:i+occlusion_size, j:j+occlusion_size] = 0
            output = F.softmax(model(occluded_image), dim=1)[0, target_class].item()
            occlusion_map[i:i+occlusion_size, j:j+occlusion_size] = base_output - output
    return occlusion_map

# Compute and visualize occlusion maps for the same 10 images
fig, axes = plt.subplots(2, 5, figsize=(15, 6))
for ax, idx in zip(axes.flatten(), indices):
    img, label = train_dataset[idx]
    occ_map = occlusion_analysis(model_resnet, img, worst_class)
    ax.imshow(occ_map, cmap='viridis')
    ax.axis('off')
plt.suptitle('Occlusion Analysis Maps for Worst-performing Class')
plt.tight_layout()
plt.show()
```

### (3f) Proposed Approach to Remove Backdoor

Based on our observations from the saliency maps and occlusion analysis, a potential backdoor might be identified by anomalous activations in specific regions. One approach to remove the backdoor is to augment or modify the training data by either removing or re-labeling the suspicious samples. For instance, using anomaly detection on the saliency maps may help identify and remove images that consistently show unexpected activation patterns.

This can be combined with data augmentation to retrain the model on a cleaner dataset.

```python
import numpy as np
from torchvision import transforms
from torch.utils.data import Subset
from torchvision.datasets import ImageFolder

def compute_anomaly_score(image):
    # compute saliency map for the given image and worst_class using the original resnet model
    saliency_map = compute_saliency(model_resnet, image, worst_class)
    # use the mean saliency as the anomaly score
    return np.mean(saliency_map)

# Compute anomaly scores on a subset of the training dataset for demonstration.
# Warning: running on the full dataset may be time consuming.
anomaly_scores = []
for idx in range(len(train_dataset)):
    image, label = train_dataset[idx]
    score = compute_anomaly_score(image)
    anomaly_scores.append(score)

anomaly_scores = np.array(anomaly_scores)
# Set threshold to flag the top 10% samples as potential backdoor samples.
threshold = np.percentile(anomaly_scores, 90)
print("Anomaly threshold (90th percentile):", threshold)

# Identify clean samples (with anomaly score below threshold)
indices_clean = np.where(anomaly_scores < threshold)[0]
print("Removed samples:", len(train_dataset) - len(indices_clean))
print("Clean training samples:", len(indices_clean))

# Create a modified training dataset by removing suspicious samples
modified_train_dataset = Subset(train_dataset, indices_clean)

# Optionally, combine the original transform with additional augmentation
aug_transform = transforms.Compose([
    transforms.RandomHorizontalFlip(),
    transforms.RandomRotation(10)
])
# Combine with the original transform (assumed to be data_transform)
combined_transform = transforms.Compose([
    data_transform,
    aug_transform
])
# Recreate the ImageFolder with the combined transform and filter with indices
modified_train_dataset_aug = Subset(
    ImageFolder(root=train_dataset.root, transform=combined_transform),
    indices_clean
)

print("Modified dataset size (with augmentation):", len(modified_train_dataset_aug))
```

### (3g) Retraining on Modified Data

After modifying the training data to remove potential backdoor samples, we retrain our ResNet-18 classifier and evaluate the new results on the test set.

```python
# Create DataLoader for modified dataset
modified_train_loader = DataLoader(modified_train_dataset_aug, batch_size=32, shuffle=True)

# Reinitialize the model for retraining
model_resnet_modified = resnet18(pretrained=False)
model_resnet_modified.fc = nn.Linear(num_ftrs, len(train_dataset.classes))
model_resnet_modified.to(device)

optimizer_modified = torch.optim.Adam(model_resnet_modified.parameters(), lr=1e-4)
criterion_modified = nn.CrossEntropyLoss()

num_epochs_mod = 15
for epoch in range(num_epochs_mod):
    model_resnet_modified.train()
    running_loss = 0.0
    for images, labels in modified_train_loader:
        images, labels = images.to(device), labels.to(device)
        optimizer_modified.zero_grad()
        outputs = model_resnet_modified(images)
        loss = criterion_modified(outputs, labels)
        loss.backward()
        optimizer_modified.step()
        running_loss += loss.item() * images.size(0)
    epoch_loss = running_loss / len(modified_train_loader.dataset)
    print(f"Modified Model - Epoch {epoch+1}/{num_epochs_mod}, Loss: {epoch_loss:.4f}")

    # Evaluate on validation set
    model_resnet_modified.eval()
    correct = 0
    total = 0
    with torch.no_grad():
        for images, labels in val_loader:
            images, labels = images.to(device), labels.to(device)
            outputs = model_resnet_modified(images)
            _, preds = torch.max(outputs, 1)
            total += labels.size(0)
            correct += (preds == labels).sum().item()
    print(f"Modified Model - Validation Accuracy: {100 * correct / total:.2f}%")
```

```python
import matplotlib.pyplot as plt
import numpy as np

def plot_predictions(model, data_loader, class_names, device):
    """
    Plot a batch of images with their predicted and true labels.

    Parameters:
        model (torch.nn.Module): The trained model.
        data_loader (DataLoader): DataLoader for the dataset to inspect.
        class_names (list): List of class names.
        device (torch.device): The device (CPU or GPU) to use.
    """
    model.eval()  # Set the model to evaluation mode
    images, labels = next(iter(data_loader))  # Get a batch of images and labels
    images, labels = images.to(device), labels.to(device)

    with torch.no_grad():
        outputs = model(images)
        _, preds = torch.max(outputs, 1)  # Get the predicted class indices

    # Move images and labels back to CPU for visualization
    images = images.cpu()
    labels = labels.cpu()
    preds = preds.cpu()

    # Plot the images with predicted and true labels
    fig, axes = plt.subplots(4, 8, figsize=(16, 8))  # Adjust grid size as needed
    axes = axes.flatten()
    for i, ax in enumerate(axes):
        if i >= len(images):  # Stop if there are fewer images than grid slots
            break
        img = images[i].permute(1, 2, 0).numpy()  # Convert from (C, H, W) to (H, W, C)
        img = np.clip(img, 0, 1)  # Ensure pixel values are in [0, 1]
        ax.imshow(img)
        ax.axis('off')
        ax.set_title(f"True: {class_names[labels[i]]}\nPred: {class_names[preds[i]]}", fontsize=8)
    plt.tight_layout()
    plt.show()

# Example usage
plot_predictions(
    model=model_resnet_modified,
    data_loader=val_loader,  # Use validation loader for inspection
    class_names=train_dataset.classes,  # Replace with your class names
    device=device
)
plot_predictions(
    model=model_resnet,
    data_loader=val_loader,  # Use test loader for inspection
    class_names=train_dataset.classes,  # Replace with your class names
    device=device
)
```

### (3h) Saliency Maps and Occlusion Analysis for the Modified Model

We generate class saliency maps and occlusion analysis maps for the same 10 training images using the modified model. This helps us compare and discuss any differences in the model's focus after removing the backdoor.

```python
# Compute saliency maps with the modified model
def compute_saliency_modified(model, image, target_class):
    model.eval()
    image = image.unsqueeze(0).to(device)
    image.requires_grad_()
    output = model(image)
    score = output[0, target_class]
    score.backward()
    saliency, _ = torch.max(image.grad.data.abs(), dim=1)
    return saliency.squeeze().cpu().numpy()

fig, axes = plt.subplots(2, 5, figsize=(15, 6))
for ax, idx in zip(axes.flatten(), indices):
    img, label = train_dataset[idx]
    saliency_mod = compute_saliency_modified(model_resnet_modified, img, worst_class)
    ax.imshow(saliency_mod, cmap='hot')
    ax.axis('off')
plt.suptitle('Modified Model: Class Saliency Maps for Worst-performing Class')
plt.tight_layout()
plt.show()

# Occlusion analysis with the modified model
def occlusion_analysis_modified(model, image, target_class, occlusion_size=10):
    model.eval()
    image = image.unsqueeze(0).to(device)
    base_output = F.softmax(model(image), dim=1)[0, target_class].item()
    h, w = image.shape[2], image.shape[3]
    occlusion_map = np.zeros((h, w))
    for i in range(0, h, occlusion_size):
        for j in range(0, w, occlusion_size):
            occluded_image = image.clone()
            occluded_image[:, :, i:i+occlusion_size, j:j+occlusion_size] = 0
            output = F.softmax(model(occluded_image), dim=1)[0, target_class].item()
            occlusion_map[i:i+occlusion_size, j:j+occlusion_size] = base_output - output
    return occlusion_map

fig, axes = plt.subplots(2, 5, figsize=(15, 6))
for ax, idx in zip(axes.flatten(), indices):
    img, label = train_dataset[idx]
    occ_map_mod = occlusion_analysis_modified(model_resnet_modified, img, worst_class)
    ax.imshow(occ_map_mod, cmap='viridis')
    ax.axis('off')
plt.suptitle('Modified Model: Occlusion Analysis Maps for Worst-performing Class')
plt.tight_layout()
plt.show()
```
